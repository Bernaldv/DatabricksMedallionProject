# 1. Eliminacion de ambiente


```python
# No usar dbutils.widgets.removeAll(): dentro de un job borraría los parámetros que envía el workflow.
dbutils.widgets.text("env", "dev", "Ambiente (dev | prod)")
dbutils.widgets.text("catalog_base", "catalog_pr_db", "Prefijo del catálogo")

 
env = dbutils.widgets.get("env").strip().lower()
catalog = f"{dbutils.widgets.get('catalog_base').strip()}_{env}"
 
assert env in ("dev", "prod"), f"env inválido: {env}"

```

Eliminacion de tablas

```python

tablas_bronze = ["circuits", "races", "constructors", "drivers", "results", "status", "driver_standings"]
tablas_silver = ["dim_circuits", "dim_drivers", "dim_constructors", "dim_status", "dim_races",
                  "fact_results", "fact_driver_standings"]
tablas_gold = ["driver_career", "constructor_career", "season_champions",
               "drivers_without_title", "constructor_standings", "constructor_dominance"]
 
for schema, tablas in [("bronze", tablas_bronze), ("silver", tablas_silver), ("gold", tablas_gold)]:
    for tabla in tablas:
        spark.sql(f"DROP TABLE IF EXISTS `{catalog}`.`{schema}`.`{tabla}`")
        print(f"Eliminada: {catalog}.{schema}.{tabla}")
```

Elimina esquema

```python
for schema in ["bronze", "silver", "gold"]:
    spark.sql(f"DROP SCHEMA IF EXISTS `{catalog}`.`{schema}` CASCADE")
    print(f"Schema eliminado: {catalog}.{schema}")
```

Elimina catalogo

```python
spark.sql(f"DROP CATALOG IF EXISTS `{catalog}` CASCADE")
print(f"Catálogo eliminado: {catalog}")
```
