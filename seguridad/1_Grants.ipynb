```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

```python
dbutils.widgets.text("env", "dev", "Ambiente (dev | prod)")
dbutils.widgets.text("catalog_base", "catalog_pr_db", "Prefijo del catálogo")
dbutils.widgets.text("schema", "gold", "Schema a otorgar")
dbutils.widgets.text("read_group", "dbrs_visualization", "Grupo/correo con acceso de lectura")

env = dbutils.widgets.get("env").strip().lower()
catalog = f"{dbutils.widgets.get('catalog_base').strip()}_{env}"
schema = dbutils.widgets.get("schema").strip()
read_group = dbutils.widgets.get("read_group").strip()

assert read_group, "read_group es obligatorio"
grupo = f"{read_group}_{env}"

print(f"catalog={catalog} | schema={schema} | grupo={grupo}")
```

```python
spark.sql(f"GRANT USE CATALOG ON CATALOG `{catalog}` TO `{grupo}`")
spark.sql(f"GRANT USE SCHEMA, SELECT ON SCHEMA `{catalog}`.`{schema}` TO `{grupo}`")
print(f"OK: USE CATALOG + SELECT en {catalog}.{schema} otorgado al grupo {grupo}")
```
